```python
import os
import numpy as np
import torch
from torch.utils.data import Dataset, DataLoader
from model import GPT, GPTConfig
# Set random seeds for reproducibility
seed = 1337
torch.manual_seed(seed)
torch.cuda.manual_seed_all(seed)
```

```python
# Load dataset
DATA_DIR = "Transformer/nanoGPT/data/shakespeare"
train_data = np.memmap(os.path.join(DATA_DIR, "train.bin"), dtype=np.uint16, mode="r")
val_data = np.memmap(os.path.join(DATA_DIR, "val.bin"), dtype=np.uint16, mode="r")
print(f"Training data shape: {train_data.shape}")
print(f"Validation data shape: {val_data.shape}")
```

```python
# Train GPT on the above dataset, don't forget to track training loss, validation loss, and wall-clock time
# Don't repeat the same code, just use the train.py and sample.py in nanoGPT
# --- 2. Model config (same defaults as train.py) ---
block_size = 256
batch_size = 64
device = "cuda" if torch.cuda.is_available() else "cpu"

config = GPTConfig(
    vocab_size=50304,  # GPT-2 tokenizer default vocab
    block_size=block_size,
    n_layer=4,
    n_head=4,
    n_embd=128,
)
model = GPT(config)
model = model.to(device)
print(f"Model initialized: {sum(p.numel() for p in model.parameters()):,} parameters")
```
